```python
import matplotlib.pyplot as plt
import numpy as np

def plot_multi_config_experiments(configurations):
    plt.figure(figsize=(20, 8))
    
    for i, (config_name, values) in enumerate(configurations):
        num_experiments = len(values)
        x = range(1, num_experiments + 1)
        
        plt.plot(x, values, 'o-', label=f'{config_name} (Experiments)', alpha=0.5)
        
        # Calculate and plot average
        avg = np.mean(values)
        plt.axhline(y=avg, linestyle='--', color=plt.gca().lines[-1].get_color(),
                    label=f'{config_name} (Average: {avg:.4f})')
    
    plt.xlabel('Experiment Number')
    plt.ylabel('Balanced Accuracy')
    plt.title('Experiment Results')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.grid(True, linestyle=':', alpha=0.7)
    plt.show()

# Example usage:
configurations = [
    ('Reg. Euclidean', normal),
    ('Polyak Euclidean', normal_polyak),
    ('Polyak Euclidean + Cosine', euclidean_cosine),
    ('Polyak Euclidean + Chebyshev', euclidean_chebyshev),
    ('Polyak Euclidean + Cosine + Chebyshev', euclidean_cosine_chebyshev_polyak),
    ('Polyak Euclidean + Cosine + Chebyshev + Wasserstein', euclidean_cosine_chebyshev_wasserstein_polyak),
    ('Polyak Chebyshev + Wasserstein',chebyshev_wasserstein_polyak),
    ('Polyak Euclidean + Chebyshev + Wasserstein + Polyak', euclidean_chebyshev_wasserstein_polyak),
    ('Polyak Euclidean + Chebyshev + Wasserstein + Polyak 2', euclidean_chebyshev_wasserstein_polyak2)
]

plot_multi_config_experiments(configurations)
```
