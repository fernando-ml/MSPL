```python
import wandb
import json
import os
import pandas as pd
import numpy as np
from scipy import stats
from fractions import Fraction

from matplotlib import pyplot as plt
import seaborn as sns
```

```python
def get_ci(df):
    results = {}
    print(df.columns)
    for metric in df.columns[1:]:
        mean = df[metric].mean()
        std = df[metric].std(ddof=1)  
        n = df[metric].count()
        stderr = std / np.sqrt(n)
        t_score = stats.t.ppf(0.975, df=n-1)  # 95% CI (two-sided) with t-distribution
        ci_half_width = t_score * stderr
        results[metric] = (np.round(mean, 4), np.round(ci_half_width, 4))
    return results
```

```python
table_name = 'experiment_summary_table'

data_array = []
runs_data_non_dl = {}
for run in runs:
    print(run.name)
    row = [
        run.name,
        run.config['dataset'],
        *run.tags,
        run.config['model_type'],
    ]
    for artifact in run.logged_artifacts():
        if 'table' in artifact.name:
            print(artifact.name)
            table_dir  = artifact.download()
            table_path = os.path.join (table_dir, table_name) + '.table.json' 
            with open(table_path, 'r') as f:
                data = json.load(f)
            df = pd.DataFrame(data["data"], columns=data["columns"])
            runs_data_non_dl[run.name] = df
            confidence_intervals = get_ci(df)
            for val in confidence_intervals.values():
                row.append(val)
    data_array.append(row)
```

```python
api = wandb.Api()

# Specify the project path
project_path = "mspl-paper/MSPL" # Replace with your entity and project name

# Get all runs in the project
runs = api.runs(
    path=project_path,
    filters={"state": "finished"},
    )

print(f"Found {len(runs)} runs in project '{project_path}':")
for run in runs:
    # Access attributes for each run
    print(f"  - Run Name: {run.name}, ID: {run.id}, State: {run.state}") #[4]
    break
    # You can add filters here, e.g., based on run.config, run.summary, run.tags
    # if 'specific_tag' in run.tags:
    #     print(f"    Found run with specific tag: {run.id}") #[5]
```

```python
def get_weights_from_name(name: str) -> str:
    name = name.split("_")
    weights = []
    for n in name:
        a = n.split("-")
        if a[0] in ['euclidean', 'chebyshev', 'cosine', 'wasserstein',]:
            weights.append(a[1])
    return weights
```

```python
columnas = [
    'name',
    'dataset',
    'polyak',
    'prototypical',
    'polyak_decay',
    'euclidean',
    'chebyshev',
    'cosine',
    'wasserstein',
    'val_accuracy_ci',
    'val_loss_ci',
    'val_f1_ci',
    'val_AUPRC_ci',
    'val_MCC_ci',
    'training_time_ci'
]
```

```python
table_name = 'experiment_summary_table'

data_array_dl = []
runs_data_dl = {}
for run in runs:
    print(run.name)
    row = [
        run.name,
        run.config['dataset'],
        *run.tags,
        run.config.get('polyak_decay', None),
        *get_weights_from_name(run.name),
    ]
    for artifact in run.logged_artifacts():
        if 'table' in artifact.name:
            print(artifact.name)
            table_dir  = artifact.download()
            table_path = os.path.join (table_dir, table_name) + '.table.json' 
            with open(table_path, 'r') as f:
                data = json.load(f)
            df = pd.DataFrame(data["data"], columns=data["columns"])
            runs_data_dl[run.name] = df
            confidence_intervals = get_ci(df)
            for val in confidence_intervals.values():
                row.append(val)
    data_array_dl.append(row)
```

```python
final_dl = pd.DataFrame(data_array_dl, columns=columnas)
```

```python
def safe_float(x):
    try:
        if isinstance(x, (float, int)):
            return float(x)
        return float(Fraction(str(x)))
    except (ValueError, ZeroDivisionError, TypeError):
        return np.nan
```

```python
numeric_cols = [final_dl[col].apply(safe_float) for col in ['euclidean', 'chebyshev', 'cosine', 'wasserstein']]
```

```python
final_dl['regular'] = (
    (final_dl['euclidean'] == '1') | 
    (final_dl['chebyshev'] == '1') | 
    (final_dl['cosine'] == '1') | 
    (final_dl['wasserstein'] == '1')
)

final_dl['val_accuracy'] = final_dl['val_accuracy_ci'].apply(lambda x: x[0])
final_dl['val_f1'] = final_dl['val_f1_ci'].apply(lambda x: x[0])
final_dl['val_AUPRC'] = final_dl['val_AUPRC_ci'].apply(lambda x: x[0])
```

```python
violin_dfs = []
for name in results['name'].unique():
    run_df  = runs_data_dl[name]
    run_df['dataset'] = results.loc[results['name'] == name]['dataset'].unique()[0]
    run_df['EMA'] = results.loc[results['name'] == name]['EMA'].unique()[0]
    run_df['method'] = results.loc[results['name'] == name]['method'].unique()[0]
    violin_dfs.append(run_df)

for name in results_non_dl['name'].unique():
    run_df  = runs_data_non_dl[name]
    run_df['dataset'] = results_non_dl.loc[results_non_dl['name'] == name]['dataset'].unique()[0]
    run_df['EMA'] = False
    run_df['method'] = 'Non-DL'
    violin_dfs.append(run_df)

violin_dfs = pd.concat(violin_dfs)
```

```python
# Hue colors
hue_colors = {True: "#8A09FF", False: "#FFC600"}

def plot_violin(data, metric='val_accuracy', **kwargs):
    sns.color_palette("pastel")
    # Create a FacetGrid with one subplot per dataset
    g = sns.FacetGrid(data, col="dataset", col_wrap=4, sharey=False, sharex=True, height=4, aspect=1.2)
    # Map the violinplot to each facet
    g.map(
        sns.violinplot,
        "method",
        metric,
        "EMA",  # Using hue for EMA
        split=True,
        palette=sns.color_palette("pastel",2),  # Use the defined custom colors
        hue_order=[False, True],
        order=['MSPL', 'reg. prototypical', 'Non-DL'],
        **kwargs
    )
    # Add legend and adjust titles/labels
    g.add_legend(title="EMA")
    g.set_axis_labels("Method", metric.replace('_', ' ').title())  # Dynamic y-label based on metric
    g.set_titles("{col_name}")  # Use dataset names as subplot titles
    g.tick_params(axis='both', labelsize=10)
    
    for ax in g.axes.flatten():
        ax.set_ylim(top=1)

    # Show plot
    plt.subplots_adjust(top=0.9)  # More reasonable spacing
    #plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_combined_violin_metrics(data, metrics=['val_accuracy', 'val_f1', 'val_AUPRC'], **kwargs):
    """
    Create a combined plot with multiple metrics in rows and datasets in columns
    """
    # Get unique datasets
    datasets = data['dataset'].unique()
    n_datasets = len(datasets)
    n_metrics = len(metrics)
    
    # Create the figure
    fig, axes = plt.subplots(nrows=n_metrics, ncols=n_datasets, 
                            figsize=(5*n_datasets, 4*n_metrics))
    
    # Ensure axes is 2D array even for single row/column
    if n_metrics == 1:
        axes = axes.reshape(1, -1)
    if n_datasets == 1:
        axes = axes.reshape(-1, 1)
    
    # Define metric labels for y-axis
    metric_labels = {
        'val_accuracy': 'Val Accuracy',
        'val_f1': 'Val F1',
        'val_AUPRC': 'Val AUPRC'
    }
    
    # Plot for each combination of metric and dataset
    for i, metric in enumerate(metrics):
        for j, dataset in enumerate(datasets):
            # Filter data for current dataset
            dataset_data = data[data['dataset'] == dataset]
            
            # Create violin plot
            sns.violinplot(
                data=dataset_data,
                x="method",
                y=metric,
                hue="EMA",
                split=True,
                palette=sns.color_palette("pastel", 2),
                hue_order=[False, True],
                order=['MSPL', 'reg. prototypical', 'Non-DL'],
                ax=axes[i, j],
                **kwargs
            )
            
            # Set titles and labels
            if i == 0:  # Top row
                axes[i, j].set_title(dataset)
            
            if j == 0:  # Left column
                axes[i, j].set_ylabel(metric_labels.get(metric, metric.replace('_', ' ').title()),
                                     fontsize=14)
            else:
                axes[i, j].set_ylabel("")
            
            if i == n_metrics - 1:  # Bottom row
                axes[i, j].set_xlabel("Method", fontsize=14)
                axes[i, j].tick_params(axis='x', rotation=45)
            else:
                axes[i, j].set_xlabel("")
                axes[i, j].set_xticklabels([])
            
            # Set y-axis limit
            axes[i, j].set_ylim(top=1.0)
            
            # Increase tick size
            axes[i, j].tick_params(axis='both', labelsize=12)
            
            # Remove individual legends (we'll add one global legend)
            axes[i, j].get_legend().remove()
    
    # Add a single legend at the bottom with two columns
    handles, labels = axes[0, 0].get_legend_handles_labels()
    fig.legend(handles, labels, title="EMA", 
              bbox_to_anchor=(0.5, -0.05), loc='upper center', 
              ncol=2, frameon=True)
    
    plt.tight_layout()
    plt.subplots_adjust(bottom=0.1)  # Make room for legend
    return fig, axes

# Alternative approach: Create separate function that doesn't close figures
def plot_violin_v3(data, metric='val_accuracy', **kwargs):
    """Modified version that returns figure without closing it"""
    g = sns.FacetGrid(data, col="dataset", col_wrap=4, sharey=False, sharex=True, height=4, aspect=1.2)
    g.map(
        sns.violinplot,
        "method",
        metric,
        "EMA",
        split=True,
        palette=sns.color_palette("pastel", 2),
        hue_order=[False, True],
        order=['MSPL', 'reg. prototypical', 'Non-DL'],
        **kwargs
    )
    g.add_legend(title="EMA")
    g.set_axis_labels("Method", metric.replace('_', ' ').title())
    g.set_titles("{col_name}")
    g.tick_params(axis='both', labelsize=10)
    for ax in g.axes.flatten():
        ax.set_ylim(top=1)
    return g
```
